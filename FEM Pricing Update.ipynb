Pricing an American Put Option using the Finite Element Method (FEM) and PSOR

The Black-Scholes PDE for option pricing is well-known, but for American options, there's no closed-form solution. So:

1. iscretize the PDE using the Finite Element Method (FEM) in log-space.

2. Solve the inequality constraint using Projected Successive Over-Relaxation (PSOR: a numerical method used to solve Linear Complementarity Problems (LCPs)).

```python
# FEM for American put in a BS market
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Parameters
N = 400
M = 400
R = 3
sigma = 0.3
r = 0.05
K = 1
T = 1
omega = 1.2

h = 2 * R / (N + 1)
k = T / M
xk = np.linspace(-R, R, N + 2)

# Source term f
f = np.zeros(N)

# Assemble matrices M and A
M_matrix = np.zeros((N, N))
A = np.zeros((N, N))

for i in range(N):
    M_matrix[i, i] = 2/3 * h
    A[i, i] = sigma**2 / h + 2 * r * h / 3
    for j in range(N):
        if abs(i - j) == 1:
            M_matrix[i, j] = 1/6 * h
            A[i, j] = -sigma**2 / (2*h) + (j - i) * (sigma**2 / 2 - r) / 2 + (r * h) / 6

B = M_matrix + k * A

# PSOR solver
def psor(A, b, c, x0, omega=1.2, tol=1e-6, max_iter=10000):
    N = len(b)
    xk_curr = x0.copy()
    xk_next = x0.copy()
    it = 0

    while it < max_iter:
        for i in range(N):
            sum1 = A[i, :i] @ xk_next[:i]
            sum2 = A[i, i+1:] @ xk_curr[i+1:]
            xk_next_tilde = (b[i] - sum1 - sum2) / A[i, i]
            xk_next_relaxed = xk_curr[i] + omega * (xk_next_tilde - xk_curr[i])
            xk_next[i] = max(c[i], xk_next_relaxed)

        res = A @ xk_next - b
        if np.linalg.norm(np.minimum(xk_next - c, res), ord=np.inf) < tol:
            break

        xk_curr[:] = xk_next
        it += 1

    return xk_next, it

# Solve PDE with PSOR
def gput(x, K):
    return np.maximum(K - np.exp(x), 0)

x = xk
S = np.exp(x)

option_value = gput(x[1:-1], K)
c = gput(x[1:-1], K)

its = np.zeros(M)

for m in range(M):
    F = k * f + M_matrix @ option_value
    option_value, it = psor(B, F, c, option_value, omega=omega)
    its[m] = it

print("Average PSOR iterations:", np.mean(its))

# Plotting
def bs_formula_P(r, sigma, s, t, K):
    t = np.maximum(t, 1e-10)
    d1 = (np.log(s / K) + (r + 0.5 * sigma ** 2) * t) / (sigma * np.sqrt(t))
    d2 = d1 - sigma * np.sqrt(t)
    return -s * norm.cdf(-d1) + K * np.exp(-r * t) * norm.cdf(-d2)

x_int = x[1:-1]
S_int = np.exp(x_int)
ind = (x_int > -R / 2) & (x_int < R / 2)

plt.plot(S_int[ind], gput(x_int[ind], K), 'k--', label='Payoff', linewidth=2)
plt.plot(S_int[ind], option_value[ind], 'b-', label='American Put Value', linewidth=2)
plt.plot(S_int[ind], bs_formula_P(r, sigma, S_int[ind], T, K), 'r--', label='European Put Value', linewidth=2)
plt.xlabel('Stock price s')
plt.ylabel('Option value')
plt.title('American vs European Put Option')
plt.legend()
plt.grid(True)
plt.show()
```
